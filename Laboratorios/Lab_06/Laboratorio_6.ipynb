# EEG_1 Basal


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import butter, sosfiltfilt

# Reconfiguración de opensignalsreader
import opensignalsreader.transfer_functions.bit as bit
bit._check_ranges = lambda *args, **kwargs: None

from opensignalsreader import OpenSignalsReader

acq = OpenSignalsReader('EEG_1.txt')
fs = acq.sampling_rate  # Obtiene frecuencia de muestreo
raw = np.asarray(acq.signal('EEG'), dtype=float)

if raw.ndim == 1:
    raw = raw[:, None]
elif raw.shape[0] < raw.shape[1]:
    raw = raw.T

t = np.arange(raw.shape[0]) / fs

# Filtro pasa banda (0.5 Hz - 40 Hz)
def butter_bandpass(lowcut, highcut, fs, order=4):
    return butter(order, [lowcut, highcut], btype='band', fs=fs, output='sos')

# aplicacion del filtro
sos = butter_bandpass(lowcut=0.5, highcut=40.0, fs=fs, order=4)
filtered_signal = sosfiltfilt(sos, raw, axis=0)

n = raw.shape[1]

plt.figure(figsize=(12, 6))
# señal original
plt.subplot(211)
plt.plot(t, raw, lw=0.5)
plt.title('Señal original')
plt.ylabel('EEG (µV)')
plt.grid(True)
# señal filtrada
plt.subplot(212)
plt.plot(t, filtered_signal, lw=0.5, color = 'tab:orange')
plt.title('Señal filtrada (0.5–40 Hz)')
plt.xlabel('Tiempo (s)')
plt.ylabel('EEG (µV)')
plt.grid(True)

plt.tight_layout()
plt.show()
```
