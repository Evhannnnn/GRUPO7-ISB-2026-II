# **𝐄𝐂𝐆 𝐞𝐧 𝐫𝐞𝐩𝐨𝐬𝐨**
## DERIVACIÓN 1 (D1)

```python
# GRAFICA DE LA SEÑAL
import matplotlib.pyplot as plt
from opensignalsreader import OpenSignalsReader
import numpy as np
from scipy.signal import butter, filtfilt


acq1 = OpenSignalsReader('ECG_reposo_D1.txt')
raw_signal = acq1.signal('ECG')  # Extrae el vector de datos
fs = acq1.sampling_rate  # Obtiene la frecuencia de muestreo


# filtro pasa banda (0.5 Hz - 40 Hz)
def butter_bandpass(lowcut, highcut, fs, order=4):
    nyq = 0.5*fs
    low = lowcut/nyq
    high = highcut/nyq
    b, a = butter(order, [low, high], btype='band')
    return b, a


# aplicación del filtro a la señal
b, a = butter_bandpass(lowcut=0.5, highcut=40.0, fs=fs, order=4)
filtered_signal = filtfilt(b, a, raw_signal)

time = np.arange(len(raw_signal)) / fs

plt.figure(figsize=(12, 6))

# grafica señal original
plt.subplot(2, 1, 1)
plt.plot(time, raw_signal, color='tab:purple')
plt.title('Señal ECG Original (Raw)')
plt.ylabel('CH1 - ECG (RAW)')
plt.grid(True)

# grafica señal filtrada
plt.subplot(2, 1, 2)
plt.plot(time, filtered_signal, color='tab:pink')
plt.title('Señal ECG Filtrada (Pasa Banda 0.5 - 40 Hz)')
plt.xlabel('Time (s)')
plt.ylabel('Amplitud')
plt.grid(True)

plt.tight_layout()
plt.show()
```
